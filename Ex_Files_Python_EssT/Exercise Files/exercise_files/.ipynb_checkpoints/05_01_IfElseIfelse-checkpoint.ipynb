### Else statements

```python
for n in range(1, 101):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
```

### Elif statements

```python
n = 12
print ("fizz" if n % 3 == 0 else n)

fizBuzz = "Fizz" if n % 3 ==0 else n

"Fizz" if n % 3 ==0 else "Buzz" if n % 5 == 0 else n
```

```python
a = 44
b = 33
if a > b:
    print("I like you")
elif a< b:
    print("I don't like you")
else:
    print("I don't know")
```

### Single Line if statements

```python
age = int(input("Enter your age: "))

if age < 13:
    print("You are a child.")
elif age < 20:
    print("You are a teenager.")
elif age < 65:
    print("You are an adult.")
else:
    print("You are a senior.")
```
