```python
import pandas as pd

df = pd.read_csv(r"C:\Users\kanza\Downloads\NISPUF17.csv", nrows=100)

def proportion_of_education():
    educ = df['EDUC1'].dropna()
    total = len(educ)

    proportions = {
        "less than high school": (educ == 1).sum() / total,
        "high school": (educ == 2).sum() / total,
        "more than high school but not college": (educ == 3).sum() / total,
        "college": (educ == 4).sum() / total,
    }
    return proportions
print(proportion_of_education())
```

```python
import pandas as pd
def average_influenza_doses():
    df = pd.read_csv(r"C:\Users\kanza\Downloads\NISPUF17.csv", nrows=100)
    doses = df[['P_NUMFLU', 'CBF_01']].dropna()
    yes_group = doses[doses['CBF_01'] == 1]
    no_group = doses[doses['CBF_01'] == 2]
    yes_avg = yes_group['P_NUMFLU'].mean()
    no_avg = no_group['P_NUMFLU'].mean()
    return (yes_avg, no_avg)
print(average_influenza_doses())
```

```python
def corr_chickenpox():
    import scipy.stats as stats
    import numpy as np
    import pandas as pd
    
    # Load the real dataset
    df = pd.read_csv("assets/NISPUF17.csv")  # make sure the path is correct

    # Filter valid rows: HAD_CPOX must be 1 or 2, and P_NUMVRC must not be NaN
    df = df[(df['HAD_CPOX'].isin([1, 2])) & (df['P_NUMVRC'].notna())]

    # Run the correlation between the two columns
    corr, pval = stats.pearsonr(df["HAD_CPOX"], df["P_NUMVRC"])

    return corr, pval
```
