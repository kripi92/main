```python
import re

def logs():
    with open(r"C:\Users\kanza\Downloads\logdata.txt", "r") as file:
        logdata = file.read()

    # Regex pattern to extract the required fields
    pattern = r'(?P<host>\S+) - (?P<user_name>\S+) \[(?P<time>[^\]]+)\] "(?P<request>[^"]+)"'

    # Find all matches using re.finditer to get named groups
    matches = re.finditer(pattern, logdata)

    # Convert matches into a list of dictionaries
    log_list = []
    for match in matches:
        log_list.append(match.groupdict())  # Convert named groups to a dictionary

    return log_list

# Run the function and print the first few entries for verification
parsed_logs = logs()
for entry in parsed_logs[:5]:  # Print the first 5 entries
    print(entry)
len(parsed_logs)  # Print the total number of entries
```
