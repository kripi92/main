```python
people = ['Dr. Christopher Brooks', 'Dr. Kevyn Collins-Thompson', 'Dr. VG Vinod Vydiswaran', 'Dr. Daniel Romero']
def get_title_and_last_name(person):
    parts = person.split()  # Split the name into components
    title = parts[0]  # The first part is the title
    last_name = parts[-1]  # The last part is the last name
    return f"{title} {last_name}"  # Corrected f-string syntax

# Apply the function to each person in the list
titles_and_last_names = list(map(get_title_and_last_name, people))

print(titles_and_last_names)
```
